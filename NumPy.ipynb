# **arange()**
np.arange(start,stop,step)

It is similar to Pythons range(), but it returns a NumPy array

Default step value = 1


```python
a=np.arange(5)
print(a)
```

```python
a=np.arange(1,10)
print(a)
```

```python
a=np.arange(1,10,2)
print(a)
```

# **linspace()**

It creates numbers that are equally spaced between a start value and end value.

**np.linspace(start,stop,num)**

start -> starting value

end -> ending value

num -> number of numbers needed



```python
a=np.linspace(0,10,5)
print(a)
```

```python
a=np.linspace(0,10,5,endpoint=True)
print(a)
```

By default, endpoint is True. And the False endpoint deletes the last value

```python
a=np.linspace(0,10,5,endpoint=False)
print(a)
```

# **random.randint()**
random.randint(start,stop,num)

- Used to a generate random integer number in the given range

- Generates a random integer excepting the end number

```python
a=np.random.randint(1,10)
print(a)
```

```python
a=np.random.randint(1,10,5)
print(a)
```

```python
a=np.random.randint(1,10,size=(3,3))
print(a)
```
