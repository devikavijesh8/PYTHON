```python
import pandas as pd
```

```python
import numpy as np
```

```python
data=pd.Series([10,20,np.nan,40])
print(data)
```

```python
data.isnull()     # True means the value is missing
```

```python
data.notnull()
```

```python
data.isnull().sum()
```

```python
data.dropna()    #removingbnot available value
```

```python
# to fill the missing value

data.fillna(88)
```

```python
data.fillna(data.mean())  #filling with mean
```
