```python
# Query

df=pd.DataFrame({
    "Name":["Anu","Meera","John","Rahul"],
    "Age":[20,22,19,25],
    "Marks":[85,91,72,88
]})

df.query("Marks>80")
```

```python
# Query with multiple conditions

df.query("Age>20 and Marks>80")
```

```python
df.query("Age>20 or Marks>80")
```

```python
# Comparing Query with Boolean Indexing

df[df["Marks"]>80]

df.query("Marks>80")
```

```python
#  Query with variables

limit=80

df.query("Marks>@limit")
```
