```python
import pandas as pd
```

```python
names=pd.Series(["Anu","Meera","John","Rahul"])
```

```python
print(names.str.upper())
```

```python
print(names.str.lower())
```

```python
print(names.str.len())
```

```python
names=pd.Series(["Anu","   Meera"," John","Rahul "])
print(names.str.strip())
```

```python
names=pd.Series(["Anu Kumar","Meera Kumar"," John Joseph","Rahul R"])

print(names.str.replace("Kumar","Joseph"))
```

```python
names=pd.Series(["Anu","Meera","Anu Kumar","Rahul"])

print(names.str.contains("Anu"))
```

```python
print(names[names.str.contains("Anu")])
```

```python
print(names.str.startswith("An"))
```

```python
print(names.str.endswith("An"))
```

```python
data=pd.Series(["Anu,20,Kochi","Meera,21,Kottayam","John,22,Thrissur"])
print(data.str.split(","))
```

```python
print(data.str.split(",",expand=True))
```
