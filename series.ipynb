BOOLIAN INDEXING

```python
student={"Name":["Ann","Ben","John"],"Age":[20,30,40],"Marks":[40,50,75],"Place":["Kottayam","Alappuzha","Trivandrum"]}
df=pd.DataFrame(student, index=["s1","s2","s3"])
print(df)
```

```python
df[df["Age"]>35]
```

```python
# multiple conditions
df[(df["Age"]>20)]
```

```python
df[(df["Age"]>25) & (df["Marks"]>70)]
```

```python
df[(df["Age"]>25) | (df["Marks"]>70)]
```
